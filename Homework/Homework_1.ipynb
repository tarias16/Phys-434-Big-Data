```python
import numpy as np
import matplotlib.pyplot as plt
import scipy
from scipy import stats

plt.rcParams["figure.figsize"] = (15,10)
```

### Pick a central value and width for a Gaussian distribution and produce histograms of the results

```python
# Sampling 100k datapoints using norm.rvs with a central value(loc) of 12 and a width(scale) of 0.05
d = stats.norm.rvs(loc = 12, scale = 0.05, size = 100000)
```

 The scipy.stats.norm function uses the Gaussian probability density function:
$$f(x) = \frac{e^{-x^2 / 2}}{ \sqrt{2\pi}} $$

```python
# Set up the initial histogram with 50 bins and overplot the probability density func. with norm.pdf
fig, ax = plt.subplots()
ax.hist(d, 50, density = True, label = 'Gaussian Data Points')
plt.tick_params(labelsize = 18)
plt.xlim(11.8, 12.2)
# Select 1000 values of x for the pdf to create smooth curve
x = np.linspace(11.8,12.2, 1000)
ax.plot(x, stats.norm.pdf(x, loc = 12, scale = 0.05), linewidth = 8, alpha = 0.7, label = 'Probability Density')
plt.legend(loc = 0)
plt.show()
```

```python
# Produces the histogram on its own with a log y-axis
fig, ax = plt.subplots()
ax.hist(d, 50, density = True, label = 'Gaussian Data Points')
plt.tick_params(labelsize = 18)
plt.xlim(11.8, 12.2)
plt.yscale('log')
plt.legend(loc = 0)
plt.show()
```

```python
# Reimplements the pdf into the log scale plot
fig, ax = plt.subplots()
ax.hist(d, 50, density = True, label = 'Gaussian Data Points')
plt.tick_params(labelsize = 18)
plt.xlim(11.8, 12.2)
plt.yscale('log')
# Select 1000 values of x for the pdf to create smooth curve
x = np.linspace(11.8,12.2, 1000)
ax.plot(x, stats.norm.pdf(x, loc = 12, scale = 0.05), linewidth = 8, alpha = 0.7, label = 'Probability Density')
plt.legend(loc = 0)
plt.show()
```

In the logarithmic scale, the bell curve resembles a parabola which is characteristic of a Gaussian distribution as described in class.
